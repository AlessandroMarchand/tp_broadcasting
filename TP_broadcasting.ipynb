```python
# les bornes pour le tirage au sort initial
mass_max = 3.
    
x_min, x_max = -10., 10.
y_min, y_max = -10., 10.

speed_max = 1.
```

```python
def init_problem(N):
    masses=np.array([rd.uniform(10**(-43),mass_max) for i in range(N)])
    positions=np.array([[rd.uniform(x_min,x_max) for i in range(N)],[rd.uniform(y_min,y_max) for i in range(N)]])
    vitesses=np.array([[rd.uniform(0,speed_max) for i in range(N)],[rd.uniform(0,speed_max) for i in range(N)]])
    return masses,positions,vitesses

#j'aurai pu utiliser la fonction np.random.uniforme qui prend en argument les bornes pour chauque dimension
# sous forme d'un tabelau à la bonne dimension et la shape du tableau à créer

masses,positions,speeds=init_problem(10)

try:
    assert masses.shape == (10,) and positions.shape == speeds.shape == (2, 10)
    print("OK")
except:
    print("KO")
```
